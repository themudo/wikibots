```python
#Script para contar número de monumentos diferentes fotografados numa edição do WLM Portugal
import pywikibot
#from pywikibot import pagegenerators as pg
import sys
sys.path.append("/srv/paws/pwb/scripts")
import mwparserfromhell
import requests
import re
import csv
import itertools
```

```python
def parse(title):
    """Parses templates in a page using mwparserfromhell"""
    site = pywikibot.Site('commons', 'commons')
    page = pywikibot.Page(site, title)
    text = page.get()
    return mwparserfromhell.parse(text)
```

```python
#Category:Images from Wiki Loves Monuments 2026 in Portugal

#Get images from competition
#PetScan: https://petscan.wmcloud.org/?psid=30314255 (change the category to match the year of interest)
#Export as tsv (despite the name of the file below...)
filename = 'Petscan_2026.tsv'
with open(filename) as f:
    for line in itertools.islice(f, 1, None):
        data = line.split("\t")
        pagename = data[1]
        commons_pagename = "File:" + pagename
        #print(commons_pagename)
        uploader = data[12]
        page_text = parse(commons_pagename)
        #Extract ID de {{MonumentIDPortugal}} 
        for template in page_text.filter_templates():
            if template.name.matches("MonumentIDPortugal"):
                with open("WLM2026_names.txt", "a", encoding="utf-8") as g:
                    g.write(uploader)
                    g.write("\t")
                    for items in template.params:
                        g.write('%s\t' %items)
                    g.write("\n")
                    print(uploader, template.params)
#print(data, "\t", pagename, "\t", uploader)
#print(pagename)
#print(uploader)

```
